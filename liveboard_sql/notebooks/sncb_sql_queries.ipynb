```python
### Level 1 — Filtering & sorting
# List all bus routes only (i.e. filter by route_type).

cursor.execute("""
    SELECT
        *
    FROM
        routes
    WHERE
        route_type = 3
    """
        )
cursor.fetchall()
```

```python
# Find every stop whose name contains "Brussels."
cursor.execute("""
    SELECT
        stop_id, stop_name
    FROM
        stops
    WHERE
        stop_name LIKE '%Brussels%'
    """
        )
cursor.fetchall()
```

```python
# List all trips heading toward a specific trip_headsign, sorted alphabetically by trip_id.
cursor.execute("""
    SELECT
        trip_id, trip_headsign
    FROM
        trips
    WHERE
        trip_headsign = 'Bruxelles-Midi'
    ORDER BY
        trip_id
    """
        )
cursor.fetchall()
```

```python
# Find all services that run on both Saturday and Sunday.
cursor.execute("""
    SELECT
        service_id
    FROM
        calendar
    WHERE
        saturday = 1 AND sunday = 1
    """
        )
cursor.fetchall()
```
